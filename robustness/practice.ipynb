```python
def explore(d, indent=0):
    for k, v in d.items():
        print("  " * indent + f"- {k}: {type(v)}")
        if isinstance(v, dict):
            explore(v, indent + 1)

explore(ckpt)
```

```python
import torch
from omegaconf import DictConfig, ListConfig, OmegaConf
from dataclasses import is_dataclass, asdict

def explore_any(obj, indent=0, max_depth=50, max_items=50):
    prefix = "  " * indent

    if indent > max_depth:
        print(prefix + "… (max depth reached)")
        return

    # --- dict ---
    if isinstance(obj, dict):
        print(prefix + f"dict ({len(obj)} keys)")
        for i, (k, v) in enumerate(obj.items()):
            if i >= max_items:
                print(prefix + "  … (more keys)")
                break
            print(prefix + f"- {k}: {type(v).__name__}")
            explore_any(v, indent + 1, max_depth, max_items)

    # --- list / tuple ---
    elif isinstance(obj, (list, tuple)):
        print(prefix + f"{type(obj).__name__} (len={len(obj)})")
        for i, v in enumerate(obj):
            if i >= max_items:
                print(prefix + "  … (more items)")
                break
            print(prefix + f"- [{i}]: {type(v).__name__}")
            explore_any(v, indent + 1, max_depth, max_items)

    # --- torch tensor ---
    elif torch.is_tensor(obj):
        print(prefix + f"Tensor shape={tuple(obj.shape)}, dtype={obj.dtype}, device={obj.device}")

    # --- OmegaConf ---
    elif isinstance(obj, (DictConfig, ListConfig)):
        print(prefix + f"{type(obj).__name__}")
        container = OmegaConf.to_container(obj, resolve=False)
        explore_any(container, indent + 1, max_depth, max_items)

    # --- dataclass ---
    elif is_dataclass(obj):
        print(prefix + f"dataclass {type(obj).__name__}")
        explore_any(asdict(obj), indent + 1, max_depth, max_items)

    # --- fallback ---
    else:
        val = repr(obj)
        if len(val) > 120:
            val = val[:120] + "…"
        print(prefix + f"value = {val}")

```

```python
explore_any(ckpt)
```
